```python
import lmdb
import json
import pandas as pd
import numpy as np
from pathlib import Path


dfs_lmdb = {}
for data_type in ['train', 'valid', 'test']:
    env = lmdb.open(f'HumanPPI/normal/{data_type}/')
    res = []

    with env.begin() as txn:
        for key, value in txn.cursor():
            if key == b'info' or key == b'length':
                continue
            res.append(json.loads(value))
    
    env.close()

    dfs_lmdb[data_type] = pd.DataFrame(res)

df_lmdb = pd.concat(dfs_lmdb.values())

sequences = list(set(df_lmdb.seq_1) | set(df_lmdb.seq_2))
len(sequences)
```

```python
import time
import random

import pandas as pd
from esm.sdk import client
from pathlib import Path
from esm.sdk.api import (
    ESMProtein,
    LogitsConfig,
)
from esm.sdk import batch_executor


MY_TOKEN = 'uio5AXjvu5Pu3BLNU5EkA'

# model_name = "esm3-large-2024-03"
# model_name = "esmc-6b-2024-12"
model_name = "esm3-medium-2024-08"

model_name_for_file = model_name.replace('/', '-').replace(' ', '-').replace('_', '-')
f = Path(f'embeddings_{model_name_for_file}.parquet')

if f.exists():
    df_existing = pd.read_parquet(f)
    done_sequences = set(df_existing['sequence'])
else:
    df_existing = pd.DataFrame(columns=['sequence', 'embedding'])
    done_sequences = set()


def calculate_embedding(model, sequence):
    protein = ESMProtein(sequence=sequence)
    protein_tensor = model.encode(protein)
    logits_output = model.logits(
        protein_tensor, LogitsConfig(sequence=True, return_embeddings=True)
    )
    emb = logits_output.embeddings.mean(dim=1)
    return emb[0].tolist()


model = client(model_name, token=MY_TOKEN)

from torch.utils.data import DataLoader
dataloader = DataLoader(
    [s for s in sequences if s not in done_sequences],
    batch_size=40
)

for batch_sequences in dataloader:
    batch_sequences = list(set([seq for seq in batch_sequences if seq not in done_sequences]))
    if len(batch_sequences) == 0:
        continue

    print(f'calculating embeddings for {len(batch_sequences)} sequences...')
    with batch_executor(max_attempts=40) as executor:
        batch_embeddings = executor.execute_batch(user_func=calculate_embedding, model=model, sequence=batch_sequences)

    new_rows = pd.DataFrame({"sequence": batch_sequences, "embedding": batch_embeddings})

    df_existing = pd.concat([df_existing, new_rows], ignore_index=True)
    df_existing.to_parquet(f, index=False)

    for seq in batch_sequences:
        done_sequences.add(seq)

    t = 10 * (random.random() + 1)
    print(f'sleeping {t} seconds...')
    time.sleep(t)
```
